```python
print("03_leap_year")

def leap_year():
    year = int(input('Please input a year: '))

    if year % 4 == 0:
        if year % 100 == 0:
            if year % 400 == 0:
                print(f"{year} a leap year!")
            else:  # (Not divisible by 400)
                print(f"{year} not a leap year.")
        else:  # (Not divisible by 100)
            print(f"{year} a leap year!")
    else:  # (Not divisible by 4)
        print(f"{year} not a leap year.")

if __name__ == '__main__':
    main()
```
